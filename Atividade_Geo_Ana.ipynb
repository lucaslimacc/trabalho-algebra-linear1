## Introdução

Este notebook implementa funções para:
- Ler (ou definir) uma matriz quadrada;
- Calcular o **determinante** da matriz;
- Calcular a **matriz inversa**;
- Exibir os resultados de forma clara.

A matriz utilizada é baseada na matriz métrica de Schwarzschild (equação 1 do enunciado), parametrizada pelos valores de `r`, `m`, `theta` (θ) e `phi` (ϕ).

---
## 1. Importações

```python
import numpy as np
```

---
## 2. Funções Auxiliares

```python
def imprimir_matriz(matriz, nome="Matriz"):
    """
    Exibe a matriz formatada no console.

    Parâmetros:
        matriz (np.ndarray): A matriz a ser exibida.
        nome (str): Rótulo/nome da matriz para exibição.
    """
    print(f"\n{nome}:")
    print("-" * 60)
    for linha in matriz:
        # Formata cada elemento com 8 casas decimais e largura fixa
        print("  ".join(f"{val:12.6f}" for val in linha))
    print("-" * 60)
```

```python
def calcular_determinante(matriz):
    """
    Calcula o determinante de uma matriz quadrada.

    Parâmetros:
        matriz (np.ndarray): Matriz quadrada de entrada.

    Retorna:
        float: O valor do determinante.
    """
    # Verifica se a matriz é quadrada
    linhas, colunas = matriz.shape
    if linhas != colunas:
        raise ValueError("A matriz deve ser quadrada para calcular o determinante.")

    # np.linalg.det calcula o determinante via decomposição LU
    det = np.linalg.det(matriz)
    return det
```

```python
def calcular_inversa(matriz):
    """
    Calcula a matriz inversa de uma matriz quadrada.

    Parâmetros:
        matriz (np.ndarray): Matriz quadrada de entrada.

    Retorna:
        np.ndarray: A matriz inversa, ou None se a matriz for singular.
    """
    det = calcular_determinante(matriz)

    # Uma matriz só é invertível se seu determinante for diferente de zero
    if np.isclose(det, 0):
        print("A matriz é singular (determinante = 0) e não possui inversa.")
        return None

    # np.linalg.inv calcula a inversa via decomposição LU
    inversa = np.linalg.inv(matriz)
    return inversa
```

## 3. Definição da Matriz do Enunciado

A matriz abaixo corresponde à equação (1) do enunciado, com os seguintes componentes:

$$
A = \begin{pmatrix}
\gamma_{00} & 0 & 0 & 0 \\
0 & \gamma_{11}\sin\theta\cos\phi & r\cos\theta\cos\phi & -r\sin\theta\sin\phi \\
0 & \gamma_{11}\sin\theta\sin\phi & r\cos\theta\sin\phi & r\sin\theta\cos\phi \\
0 & \gamma_{11}\cos\theta & -r\sin\theta & 0
\end{pmatrix}
$$

Onde:
$$\gamma_{00} = \frac{r}{\sqrt{1 - \frac{2m}{r}}}, \qquad \gamma_{11} = \sqrt{1 - \frac{2m}{r}}^{-1}$$

```python
def construir_matriz(r, m, theta, phi):
    """
    Constrói a matriz A do enunciado para dados parâmetros físicos.

    Parâmetros:
        r     (float): Coordenada radial (deve satisfazer r > 2m).
        m     (float): Parâmetro de massa.
        theta (float): Ângulo polar em radianos.
        phi   (float): Ângulo azimutal em radianos.

    Retorna:
        np.ndarray: Matriz 4x4 construída com os parâmetros fornecidos.
    """
    # Fator de Schwarzschild: (1 - 2m/r)
    fator = 1 - (2 * m / r)

    if fator <= 0:
        raise ValueError("O parâmetro r deve ser maior que 2m (fora do horizonte de eventos).")

    # Componentes gamma conforme equações (2) e (3) do enunciado
    gamma_00 = r / np.sqrt(fator)       # equação (2)
    gamma_11 = 1 / np.sqrt(fator)       # equação (3)

    # Senos e cossenos dos ângulos
    sen_t = np.sin(theta)
    cos_t = np.cos(theta)
    sen_p = np.sin(phi)
    cos_p = np.cos(phi)

    # Montagem da matriz 4x4 conforme equação (1)
    A = np.array([
        [gamma_00,                  0,                    0,                  0           ],
        [0,        gamma_11*sen_t*cos_p,  r*cos_t*cos_p,  -r*sen_t*sen_p],
        [0,        gamma_11*sen_t*sen_p,  r*cos_t*sen_p,   r*sen_t*cos_p],
        [0,        gamma_11*cos_t,       -r*sen_t,          0            ]
    ])

    return A
```

---
## 4. Execução Principal

Defina os parâmetros `r`, `m`, `theta` e `phi` e execute o cálculo.

```python
# ============================================================
# PARÂMETROS DE ENTRADA — altere os valores conforme desejado
# ============================================================
r     = 10.0              # Coordenada radial (deve ser > 2m)
m     = 1.0               # Parâmetro de massa
theta = np.pi / 4         # Ângulo polar: 45°
phi   = np.pi / 3         # Ângulo azimutal: 60°
# ============================================================

# 1. Construir a matriz
A = construir_matriz(r, m, theta, phi)
imprimir_matriz(A, nome="Matriz A")

# 2. Calcular e exibir o determinante
det = calcular_determinante(A)
print(f"\nDeterminante de A: {det:.6f}")

# 3. Calcular e exibir a matriz inversa
A_inv = calcular_inversa(A)
if A_inv is not None:
    imprimir_matriz(A_inv, nome="Matriz Inversa A⁻¹")

# 4. Verificação: A * A⁻¹ deve ser a matriz identidade
if A_inv is not None:
    identidade = A @ A_inv
    print("\nVerificação — A × A⁻¹ (deve ser a identidade):")
    imprimir_matriz(np.round(identidade, 6), nome="A × A⁻¹")
```
